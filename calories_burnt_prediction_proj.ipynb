importing dependencies

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.model_selection import train_test_split
from xgboost import XGBRegressor
from sklearn import metrics
from sklearn.preprocessing import StandardScaler
```

```python
calorie=pd.read_csv('/content/calories.csv')
calorie.head()
```

```python
exercise=pd.read_csv('/content/exercise.csv')
exercise.head()
```

combining 2 data frames  in pandas axis=1 represent coln and axis=0 represent rows

```python
cal_data=pd.concat([exercise,calorie['Calories']],axis=1)
cal_data.head()
```

calculating rows and coln

inplace is used to change the original data and carry the replacement throughout

```python
cal_data.replace({"Gender":{'male':0,'female':1}},inplace=True)
```

```python
cal_data.head()
```

seperating features and target

```python
x=cal_data.drop(columns=['User_ID','Calories'],axis=1)
y=cal_data['Calories']
```

```python
print(x)
```

```python
print(y)
```

splitting data into training and testing data

```python
x_train,x_test,y_train,y_test=train_test_split(x,y,test_size=0.2,random_state=2)
```

model training

```python
model=XGBRegressor()
```

training  the model

```python
model.fit(x_train,y_train)
```

evaluation
prediction on test data

```python
test_data_prediction=model.predict(x_test)
print(test_data_prediction)
```

mean absolute error

```python
m=metrics.mean_absolute_error(y_test,test_data_prediction)
print(m)
```

making a predictive system

```python
input_data=(1,20,166.0,60.0,14.0,94.0,40.3)
input_data_as_numpy_array=np.array(input_data)
```

```python
input_data_reshaped=input_data_as_numpy_array.reshape(1,-1)
```

```python
prediction=model.predict(input_data_reshaped)
print("calories burnt:",prediction[-1])
```
